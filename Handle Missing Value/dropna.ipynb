```python
import pandas as pd
data = {
    "Name": ['Ram', None, 'Ghanshyam', 'Dhanshyam', 'Aditi', 'Jagdish', 'Raj', 'Simran'],
    "Age": [28, 34, None, 30, 29, 40, 25, 32],
    "Salary": [50000, 60000, 40000, 52000, 49000, 70000, 48000, 58000],
    "Performance Score": [85, 90, 78, 92, 88, 95, 80, 89]
}

df=pd.DataFrame(data)

df.isnull().sum()
print(df.head(8))
```

```python
df.dropna(axis=0, inplace=True)
print(df.head(8))
```
