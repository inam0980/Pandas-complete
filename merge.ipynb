```python
import pandas as pd

data1={
    "C_id":[1,2,3,4,5],
    "C_name":["Inam Khan","Arun","Varun","Karun","Narun"],
    "C_age":[28,28,34,22,34]
}


data2={
    "C_id":[1,2,3,4,6],
    "C_salary":[50000,60000,45000,52000,48000]
}

df1=pd.DataFrame(data1)
df2=pd.DataFrame(data2)
```

```python
Merged_df=pd.merge(df1,df2,on="C_id",  how="inner")
```

Innner Join

```python
Merged_df.head()
```

Outer Join

```python
outer_m=pd.merge(df1,df2, on="C_id", how="outer")
print(outer_m)
```

left Join

```python
merge_l=pd.merge(df1,df2, on="C_id",  how="left")
print(merge_l)
```

Rightjoin

```python
merge_r=pd.merge(df1,df2,  on="C_id", how="right")
```

```python
print(merge_r)
```

Cross Join

```python
merge_c = pd.merge(df1, df2, how="cross")
print(merge_c)
```
