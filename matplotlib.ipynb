```python
import pandas as pd
data = {
    "Name": ['Ram', None, 'Ghanshyam', 'Dhanshyam', 'Aditi', 'Jagdish', 'Raj', 'Simran'],
    "Age": [28, 34, 77, 30, 29, 40, 25, 32],
    "Salary": [150000, 60000, 45555, 52000, 49000, 70000, 48000, 58000],
    "Performance Score": [85, 90, 78, 92, 88, 95, 80, 89]
}

df=pd.DataFrame(data)
df.head()
```

```python
import matplotlib.pyplot as plt

plt.figure()
plt.plot(df["Age"], df["Salary"])
plt.xlabel("Age")
plt.ylabel("Salary")
plt.title("Age vs Salary Trend")
plt.show()

```

```python
plt.figure()
plt.scatter(df["Age"], df["Salary"])
plt.xlabel("Age")
plt.ylabel("Salary")
plt.grid()
plt.title("Age vs Salary Relationship")
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd

df["Salary"] = pd.to_numeric(df["Salary"], errors="coerce")

plt.boxplot(df["Salary"])
plt.title("Salary Box Plot")
plt.grid()
plt.show()
```
