```python
import pandas as pd
data = {
    "Name": ['Ram', None, 'Ghanshyam', 'Dhanshyam', 'Aditi', 'Jagdish', 'Raj', 'Simran'],
    "Age": [28, 34, None, 30, 29, 40, 25, 32],
    "Salary": [50000, 60000, None, 52000, 49000, 70000, 48000, 58000],
    "Performance Score": [85, 90, 78, 92, 88, 95, 80, 89]
}

df=pd.DataFrame(data)
df.head()
```

```python
df.sort_values(by=["Name","Age"],ascending=[True,True], inplace=True)
```

```python
df["Salary"]=df["Salary"].interpolate(method='linear')
```
